```python
import matplotlib.pyplot as plt
import numpy as np
from scipy.interpolate import CubicSpline

# Heights (mm)
heights = np.array([10, 20, 30, 40, 50])

# Pressure difference data (Pa)
dp_staggered = np.array([1.84, 2.15, 2.75, 3.77, 4.30])
dp_inline = np.array([1.85, 2.18, 2.78, 3.87, 4.38])

# Percentage change: (Inline - Staggered) / Staggered * 100
percent_change = ((dp_inline - dp_staggered) / dp_staggered) * 100

# Smooth interpolation
heights_smooth = np.linspace(heights.min(), heights.max(), 500)
cs_staggered = CubicSpline(heights, dp_staggered)
cs_inline = CubicSpline(heights, dp_inline)

# Plotting
fig, ax = plt.subplots(figsize=(10, 6.5))
ax.plot(heights_smooth, cs_staggered(heights_smooth), label='Staggered', color='blue', linewidth=2)
ax.plot(heights_smooth, cs_inline(heights_smooth), label='Inline', color='orange', linewidth=2)
ax.plot(heights, dp_staggered, 'o', color='blue')
ax.plot(heights, dp_inline, 'o', color='orange')

# Labels and title
ax.set_xlabel('Fin Height (mm)', fontsize=12)
ax.set_ylabel('Pressure Difference (Pa)', fontsize=12)
ax.set_title('Pressure Difference vs Fin Height for Cylindrical Pin Fin Heatsinks', fontsize=14)
ax.grid(True)
ax.legend()

# Create table data
table_data = []
for i in range(len(heights)):
    table_data.append([
        f"{heights[i]}",
        f"{dp_staggered[i]:.2f}",
        f"{dp_inline[i]:.2f}",
    ])

# Add table below plot
column_labels = ['Height (mm)', 'Staggered ΔP', 'Inline ΔP', '% Change']
table = plt.table(cellText=table_data, colLabels=column_labels, cellLoc='center',
                  loc='bottom', bbox=[0.0, -0.42, 1, 0.3])
table.auto_set_font_size(False)
table.set_fontsize(10)

plt.subplots_adjust(left=0.1, bottom=0.32)
plt.tight_layout()
plt.show()
```
