```python
import matplotlib.pyplot as plt
import numpy as np
from scipy.interpolate import make_interp_spline

# Original data
velocity = np.array([1.9, 2.88, 3.85, 4.81, 5.81])
pressure_diff_staggered = np.array([4.3, 8.5, 13.8, 19.12, 27.4])
pressure_diff_inline = np.array([4.38, 8.93, 14.86, 22.67, 31.72])

# Calculate percentage increase (Inline vs Staggered)
percent_increase = ((pressure_diff_inline - pressure_diff_staggered) / pressure_diff_staggered) * 100

# Generate a finer velocity range for smooth curve
velocity_smooth = np.linspace(velocity.min(), velocity.max(), 300)

# Create spline functions
spline_staggered = make_interp_spline(velocity, pressure_diff_staggered, k=3)
spline_inline = make_interp_spline(velocity, pressure_diff_inline, k=3)

# Evaluate spline functions
pressure_staggered_smooth = spline_staggered(velocity_smooth)
pressure_inline_smooth = spline_inline(velocity_smooth)

# Plotting
fig, ax = plt.subplots(figsize=(10, 6.5))

# Plot smooth curves
ax.plot(velocity_smooth, pressure_staggered_smooth, label='Staggered ', linewidth=2, color='blue')
ax.plot(velocity_smooth, pressure_inline_smooth, label='Inline Heat', linewidth=2, color='orange')

# Scatter original points
ax.scatter(velocity, pressure_diff_staggered, color='blue', s=40)
ax.scatter(velocity, pressure_diff_inline, color='orange', s=40)

# Labels and title
ax.set_xlabel('Velocity (m/s)', fontsize=12)
ax.set_ylabel('Pressure Difference (Pa)', fontsize=12)
ax.set_title('Pressure Difference vs Velocity for Cylindrical Pin Fin Heatsinks', fontsize=14)
ax.grid(True)
ax.legend()

# Prepare table data
table_data = []
for i in range(len(velocity)):
    table_data.append([
        f"{velocity[i]:.2f}",
        f"{pressure_diff_staggered[i]:.2f}",
        f"{pressure_diff_inline[i]:.2f}",
    ])

# Table column labels
column_labels = ['Velocity (m/s)', 'Staggered ΔP (Pa)', 'Inline ΔP (Pa)', '% Increase']

# Add table below the plot
table = plt.table(cellText=table_data, colLabels=column_labels, cellLoc='center',
                  loc='bottom', bbox=[0.0, -0.42, 1, 0.3])
table.auto_set_font_size(False)
table.set_fontsize(10)

plt.subplots_adjust(left=0.1, bottom=0.32)  # Adjust to make room for table
plt.tight_layout()
plt.show()


```
