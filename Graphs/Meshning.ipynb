```python
import matplotlib.pyplot as plt

# Data
elements = [3.5, 4, 6, 8, 10]
temperatures = [344.998, 345.35, 345.28, 345.06, 344.635]

# Plot
plt.figure(figsize=(8, 4))
plt.plot(elements, temperatures, marker='o', linestyle='-', linewidth=2)

# Labels and title
plt.title('Mesh Independence Study: Avg. Temperature vs Element Count', fontsize=12)
plt.xlabel('Mesh Elements (in lakhs)', fontsize=10)
plt.ylabel('Average Temperature (K)', fontsize=10)
plt.xticks(elements)
plt.grid(True)

# Force a visually flat line by zooming into a tiny Y-axis window
plt.ylim(344, 345)  # Or even tighter like (344.6, 345.4) to flatten more

plt.tight_layout()
plt.show()


```
