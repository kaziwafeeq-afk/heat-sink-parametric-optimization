```python
import matplotlib.pyplot as plt
import numpy as np
import pandas as pd

# Common Re
Re = np.array([8731, 13221, 17603, 22071, 26671])

# Temperature
temp_actual = np.array([344, 338.2, 335.4, 333.1, 331])
temp_our = np.array([344.87, 339.4, 336.53, 333.9, 331.4])
error_temp = np.abs((temp_actual - temp_our) / temp_actual) * 100

plt.figure()
plt.plot(Re, temp_actual, label="Authors simulations", marker='o')
plt.plot(Re, temp_our, label="Our Simulations", linestyle='--', marker='x')
plt.xlabel("Re")
plt.ylabel("Temperature (K)")
plt.title("Avg Temperature vs Re (Solid Fins)")
plt.grid()
plt.legend()
plt.show()

# Error table
df_temp = pd.DataFrame({
    "Re": Re,
    "Actual Temp (K)": temp_actual,
    "Our Temp (K)": temp_our,
    "Error (%)": np.round(error_temp, 2)
})
print(df_temp)
```

```python
import matplotlib.pyplot as plt

# Data
reynolds_number = [8731, 13221, 17603, 22071, 26671]
heat_transfer_coeff = [139.64, 156.68, 167.50, 179.69, 193.22]

# Plot
plt.figure(figsize=(8, 5))
plt.plot(reynolds_number, heat_transfer_coeff, marker='o', linestyle='-', color='b')
plt.title('Reynolds Number vs Heat Transfer Coefficient')
plt.xlabel('Reynolds Number')
plt.ylabel('Heat Transfer Coefficient (W/m²·K)')
plt.grid(True)
plt.tight_layout()
plt.show()
```
