```python
import matplotlib.pyplot as plt
import numpy as np
from scipy.interpolate import CubicSpline, make_interp_spline

# Pin Dia / Spacing Ratios
ratios = np.array([0.32, 0.56, 0.72, 0.80, 0.92])

# Pressure Difference Data (Pa)
dp_staggered = np.array([4.3, 9.96, 22.76, 40.76, 222.54])
dp_inline = np.array([4.38, 13.0, 34.2, 60.87, 376.19])

# % Change Calculation: (Inline - Staggered) / Staggered * 100
percent_change_dp = ((dp_inline - dp_staggered) / dp_staggered) * 100

# Smooth interpolation
ratios_smooth = np.linspace(ratios.min(), ratios.max(), 500)
cs_staggered = CubicSpline(ratios, dp_staggered)
cs_inline = make_interp_spline(ratios, dp_inline, k=2)

dp_staggered_smooth = cs_staggered(ratios_smooth)
dp_inline_smooth = cs_inline(ratios_smooth)

# Plotting
fig, ax = plt.subplots(figsize=(10, 6.5))

# Plot smoothed curves
ax.plot(ratios_smooth, dp_staggered_smooth, label='Staggered', color='blue', linewidth=2)
ax.plot(ratios_smooth, dp_inline_smooth, label='Inline', color='orange', linewidth=2)

# Plot original data points
ax.scatter(ratios, dp_staggered, color='blue', marker='x')
ax.scatter(ratios, dp_inline, color='orange', marker='x')

# Axis labels and title
ax.set_xlabel('Pin Diameter / Pin Spacing Ratio', fontsize=12)
ax.set_ylabel('Pressure Difference (Pa)', fontsize=12)
ax.set_title('Pressure Difference vs Pin Dia/Spacing Ratio for Cylindrical Pin Fin Heatsinks', fontsize=14)
ax.grid(True)
ax.legend()

# Tabular data
table_data = []
for i in range(len(ratios)):
    table_data.append([
        f"{ratios[i]:.2f}",
        f"{dp_staggered[i]:.2f}",
        f"{dp_inline[i]:.2f}",
    ])

# Table labels
column_labels = ['Pin Dia/Spacing', 'Staggered ΔP', 'Inline ΔP', '% Change']

# Add table below the plot
table = plt.table(cellText=table_data, colLabels=column_labels, cellLoc='center',
                  loc='bottom', bbox=[0.0, -0.42, 1, 0.3])
table.auto_set_font_size(False)
table.set_fontsize(10)

# Layout adjustments
plt.subplots_adjust(left=0.1, bottom=0.32)
plt.tight_layout()
plt.show()
```
